# Spark delta table 作成

```python
#  ライブラリインポート
import random
from datetime import date, timedelta
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    DateType,
    IntegerType,
)
```

```python
transaction_schema = StructType([
    StructField("transaction_id", StringType(), False),
    StructField("sender", StringType(), False),
    StructField("accepter", StringType(), False),
    StructField("time_stamp", DateType(), False),
    StructField("amount", IntegerType(), False),
])


# トランザクションマスタ
sender = [
    "A さん", "B さん", "C さん",
    "D さん", "E さん", "F さん", "G さん",
    "H さん", "I さん", "J さん",
    "K さん", "L さん", "M  さん",
]


accepter = [
    "A さん", "B さん", "C さん",
    "D さん", "E さん", "F さん", "G さん",
    "H さん", "I さん", "J さん",
    "K さん", "L さん", "M  さん",
]


trans = [
    {
        "transaction_id": f"P{idx + 1:03}",
        "sender": random.choice(sender),
        "accepter": random.choice(accepter),
        "time_stamp": date.today() - timedelta(
            days=random.randint(0, 29)
        ),
        "amount": random.randint(100, 5000)
    }
    for idx in range(100)
]

df_trans = spark.createDataFrame(trans, schema=transaction_schema)
df_trans.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.datasets.transaction")

display(df_trans)
```
